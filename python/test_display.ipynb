# Test Virtual Scrolling Display Hook

This notebook tests the `metal0.lanceql.display` module which provides:
- Virtual scrolling for large DataFrames
- Image URL auto-detection
- Jupyter Comm-based row fetching

```python
# Opt-in to virtual scrolling display hook
import metal0.lanceql.display
```

```python
# Test with Polars DataFrame
import polars as pl

df = pl.DataFrame({
    'id': range(1000),
    'name': [f'Item {i}' for i in range(1000)],
    'url': ['https://picsum.photos/200/300.jpg'] * 1000,
    'value': [i * 1.5 for i in range(1000)]
})

df  # Should show virtual table with image thumbnails
```

```python
# Test with PyArrow Table
import pyarrow as pa

table = pa.table({
    'id': range(500),
    'name': [f'Record {i}' for i in range(500)],
    'score': [i * 2.5 for i in range(500)]
})

table  # Virtual scrolling
```

```python
# Test with Pandas DataFrame
import pandas as pd

pdf = pd.DataFrame({
    'id': range(800),
    'category': [f'Cat {i % 10}' for i in range(800)],
    'amount': [i * 3.14 for i in range(800)]
})

pdf  # Virtual scrolling
```
